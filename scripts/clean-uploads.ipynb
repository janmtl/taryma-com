```python
import json
import glob
import pandas as pd
import shutil
import os
```

```python
for f in glob.glob("uploads/image/filename/**/thumb_*.jpg"):
    os.remove(f)
```

```python
for f in glob.glob("uploads/image/filename/**/*.jpg"):
    id = f.split('/')[-1][:-len('.jpg')]
    dest_f = f"images/{id}.jpg"
    shutil.move(f, dest_f)
```
