```python
studies = json.load(open("../data/studies.json", "r"))
images_for_studies = json.load(open("../data/images_for_studies.json", "r"))
artworks_for_studies = json.load(open("../data/artworks_for_studies.json", "r"))
artworks_filenames_and_titles = json.load(open("../data/artworks_filenames_and_titles.json", "r"))
images_filenames = json.load(open("../data/images_filenames.json", "r"))
```

```python
artworks_template = """
![{title}]("https://taryma-com-public.hel1.your-objectstorage.com/artworks-browse/{filename}")
"""
images_template = """
![Image]("https://taryma-com-public.hel1.your-objectstorage.com/images-browse/{filename}")
"""

body_template = """
# {name}

{description}

{artworks}
{images}
"""

for _, row in pd.DataFrame(studies).iterrows():
    id = str(row['id'])
    slug = row['name'].lower().replace(" ", "-")
    name = row['name']
    description = row['description']
    images_data = images_for_studies.get(id, [])
    artworks_data = artworks_for_studies.get(id, [])

    images = "".join([images_template.format(filename=images_filenames[str(image_id)]) for image_id in images_data])
    artworks = "".join([artworks_template.format(filename=artworks_filenames_and_titles[str(artwork_id)]['filename'], title=artworks_filenames_and_titles[str(artwork_id)]['title']) for artwork_id in artworks_data])

    body = body_template.format(name=name, description=description, artworks=artworks, images=images)

    with open(f"../studies/{slug}.md", "w") as f:
        f.write(body)
```
