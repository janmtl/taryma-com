```python
import pandas as pd
```

```python
df0 = pd.read_json("data/gallery.json")
df0.head()
```

```python
artworks_template = "![{title}](https://taryma-com-public.hel1.your-objectstorage.com/artworks-browse/{filename})"
body_template = """
---
title: "{category}"
---
{artworks}
"""

for category, data in df0.groupby("category"):
    category_slug = category.lower().replace(" ", "-")
    artworks = " ".join([
        artworks_template.format(
            title=row['title']  ,
            filename=row['filename']
        )
        for _, row in data.iterrows()
    ])
    
    body = body_template.format(
        category=category,
        artworks=artworks
    )
    
    print(f"category_slug: {category_slug}, number of images: {len(data)}")

    with open(f"content/gallery/{category_slug}.md", "w") as f:
        f.write(body)
```
